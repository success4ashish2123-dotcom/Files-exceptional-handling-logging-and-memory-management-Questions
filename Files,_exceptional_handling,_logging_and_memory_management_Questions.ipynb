- Compilation happens before execution.

- Generally faster execution.

Examples: C, C++, Rust.

18.>  What is the difference between os.fork() and multiprocessing in Python?

Ans.

The main difference between os.fork() and multiprocessing in Python is that os.fork() creates a new child process by duplicating the current process, whereas the multiprocessing module provides a higher-level way to create and manage multiple processes.

os.fork() is a low-level operating system function that is mainly available on Unix-like systems such as Linux and macOS. It creates a child process that initially has a copy of the parent's memory space. In contrast, multiprocessing is a Python module that provides functions and classes to create, manage, and communicate between multiple processes. It is more portable and works on Windows, Linux, and macOS. The multiprocessing module is generally easier to use and is suitable for CPU-intensive tasks, while os.fork() provides more direct control over process creation but requires more operating-system knowledge. Both approaches can be used to achieve parallel execution in Python.

18.  How would you implement a basic logging setup that logs to a file with rotation after 1MB.

The RotatingFileHandler is available in the logging.handlers module. We can set maxBytes=1MB (1,048,576 bytes) to limit the log file size to 1 MB and use backupCount to specify how many old log files should be kept. When the log file reaches the size limit, Python automatically creates a new log file and stores the previous log file as a backup. This helps prevent log files from growing too large.

import logging
from logging.handlers import RotatingFileHandler

# Create a rotating file handler
handler = RotatingFileHandler(
    "app.log",
    maxBytes=1024 * 1024,  # 1 MB
    backupCount=3
)

# Configure logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    handlers=[handler]
)

# Log messages
logging.info("Application started.")
logging.warning("This is a warning message.")
logging.error("An error occurred.")

print("Log messages recorded successfully.")

In Python, we can check whether a file is empty before reading its contents by using the os.path.getsize() function from the os module. This function returns the size of a file in bytes. If the file size is 0, it means the file is empty. If the file size is greater than 0, we can open the file in read mode ("r") and read its contents. We can also use try-except blocks to handle errors if the file does not exist.
import os

# Check whether the file exists
if os.path.exists("example.txt"):

    # Check whether the file is empty
    if os.path.getsize("example.txt") == 0:
        print("The file is empty.")

    else:
        # Open and read the file
        with open("example.txt", "r") as file:
            content = file.read()
            print("File content:")
            print(content)

else:
    print("Error: File does not exist.")

23.  Write a Python program that writes to a log file when an error occurs during file handling.

Ans.

In Python, we can use the logging module along with try-except blocks to record file-handling errors in a log file. When an error occurs while opening, reading, or writing a file, the except block catches the exception and uses logging.error() to write the error message into a log file. This helps developers track file-related errors and debug programs more easily without terminating the program unexpectedly.

import logging

# Configure logging to save errors in a log file
logging.basicConfig(
    filename="error.log",
    level=logging.ERROR,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

try:
    # Attempt to open a file
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)

except FileNotFoundError:
    logging.error("File not found while attempting to read.")
    print("Error: File does not exist.")

except PermissionError:
    logging.error("Permission denied while accessing the file.")
    print("Error: Permission denied.")

except OSError as e:
    logging.error("File handling error: %s", e)
    print("Error occurred while handling the file.")
